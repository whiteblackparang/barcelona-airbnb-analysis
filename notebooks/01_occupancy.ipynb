### Calendar 데이터로 리스팅별 점유율 계산

```python
import pandas as pd

cal = pd.read_csv('../data/raw/calendar.csv.gz', usecols=['listing_id', 'date', 'available'])
cal.head()
```

```python
cal['booked'] = (cal['available'] == 'f').astype(int)

occ = cal.groupby('listing_id').agg(
    total_days=('date', 'count'),
    booked_days=('booked', 'sum')
).reset_index()

occ['occupancy_rate'] = occ['booked_days'] / occ['total_days']
occ['occupancy_rate'].describe()
```

한계: available == 'f'는 "예약됨"과 "호스트가 임의로 막아놓음(개인사용/정비)"을 구분하지 못한다. 

occupancy_rate는 진짜 예약률이 아니라 "비어있지 않은 날짜 비율"

```python
occ.to_pickle('../data/processed/occupancy.pkl')
print(f"저장: {occ.shape[0]}개 리스팅")
```
