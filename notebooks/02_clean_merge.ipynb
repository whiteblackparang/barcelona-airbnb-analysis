```python
import pandas as pd

df = pd.read_csv('../data/raw/listings.csv.gz', low_memory=False)
occ = pd.read_pickle('../data/processed/occupancy.pkl')
```

```python
df.shape
```

```python
price_raw = df['price'].astype(str)
price_raw = price_raw.str.replace('$', '', regex=False)
price_raw = price_raw.str.replace(',', '', regex=False)
df['price'] = pd.to_numeric(price_raw, errors='coerce')
df['price'].describe()
```

```python
df = df.merge(occ[['listing_id', 'occupancy_rate']],
              left_on='id', right_on='listing_id', how='left')
df['occupancy_rate'].isnull().sum()
```

```python
df['is_new_host'] = df['hosts_time_as_host_years'] <= 1
df['is_new_host'].mean()
```

```python
df['number_of_reviews_ltm'] = df['number_of_reviews_ltm'].fillna(0)
print("instant_bookable 결측률:", 
      df['instant_bookable'].isnull().mean())
```

```python
df.to_pickle('../data/processed/master.pkl')
print("저장:", df.shape)
```
